```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv("Data extraction_postpartum MAR6 (version 1).csv")
df.columns = df.columns.str.strip()

# Target the specific column
column_name = "Data Input/independent features"

# Drop missing values and convert to lowercase
features_series = df[column_name].dropna().astype(str).str.lower()

# Split comma-separated features
features_split = features_series.str.split(',')

# Flatten the list and clean spaces
flat_features = [item.strip() for sublist in features_split for item in sublist if item.strip()]

# Count frequency
feature_counts = pd.Series(flat_features).value_counts().head(20).reset_index()
feature_counts.columns = ["Feature", "Count"]

# Plot the results
plt.figure(figsize=(10, 6))
sns.barplot(data=feature_counts, y="Feature", x="Count", palette="mako")
plt.title("Top 20 Most Frequently Used Independent Features")
plt.xlabel("Frequency")
plt.ylabel("Feature")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv("Data extraction_postpartum MAR6 (version 1).csv")
df.columns = df.columns.str.strip()

# Convert Mean age column to numeric
df["Mean Age"] = pd.to_numeric(df["Mean Age"], errors='coerce')

# Define 3 age bins
bins = [0, 25, 35, 100]
labels = ['<25', '26-35', '>35']
df['Age Group'] = pd.cut(df['Mean Age'], bins=bins, labels=labels)

# Count how many studies fall in each age group
age_group_counts = df['Age Group'].value_counts().sort_index().reset_index()
age_group_counts.columns = ['Age Group', 'Count']

# Plot
plt.figure(figsize=(8, 5))
sns.barplot(data=age_group_counts, x='Age Group', y='Count', palette='coolwarm')
plt.title('Distribution of Studies by Mean Age Group')
plt.xlabel('Mean Age Group')
plt.ylabel('Number of Studies')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("Data extraction_postpartum MAR6 (version 1).csv")
df.columns = df.columns.str.strip()

# Check if 'Best performing algorithm' exists
algorithm_col = [col for col in df.columns if "algorithm" in col.lower()]
print("Detected algorithm column:", algorithm_col)

# Assume correct column is detected (modify if needed)
algo_col = algorithm_col[0]  # e.g., "Best performing algorithm"
df_algo = df[algo_col].dropna().str.strip()

# Count frequency of each algorithm
algo_counts = df_algo.value_counts().reset_index()
algo_counts.columns = ["Algorithm", "Count"]

# Plot
plt.figure(figsize=(10, 6))
sns.barplot(data=algo_counts, y="Algorithm", x="Count", palette="cubehelix")
plt.title("Best Performing Algorithms in the Study")
plt.xlabel("Number of Studies")
plt.ylabel("Algorithm")
plt.tight_layout()
plt.show()
```

```python

from scipy.stats import skew


# Ensure correct column name
participant_col = 'No of participants'

# Drop missing or non-numeric values
df[participant_col] = pd.to_numeric(df[participant_col], errors='coerce')
df_clean = df.dropna(subset=[participant_col])

# Calculate skewness
skewness = skew(df_clean[participant_col])
print(f"Skewness: {skewness:.2f}")

# Plot the histogram
plt.figure(figsize=(10, 6))
sns.histplot(df_clean[participant_col], bins=30, kde=True, color='skyblue')
plt.title('Distribution of Number of Participants Across Studies')
plt.xlabel('Number of Participants')
plt.ylabel('Frequency')
plt.axvline(df_clean[participant_col].mean(), color='red', linestyle='dashed', linewidth=1, label='Mean')
plt.legend()
plt.tight_layout()
plt.show()
```
